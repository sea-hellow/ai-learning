## 4. 第一行真正的代码：算点东西

这是后面所有机器学习的基础——把数据放进数组，做运算。

```python
a = np.array([1, 2, 3, 4, 5])
print("数组:", a)
print("均值:", a.mean())
print("乘以 2:", a * 2)
print("平方:", a ** 2)
```

## 5. 画一张图

如果看到一条正弦曲线，说明 matplotlib 也通了。

**中文乱码说明**：这里标题用的是英文，所以不会乱码。以后画中文图需要额外设置字体，
到时候我告诉你怎么加。

```python
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi, 100)
plt.figure(figsize=(7, 3))
plt.plot(x, np.sin(x), label="sin(x)")
plt.plot(x, np.cos(x), label="cos(x)")
plt.title("My first plot")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```
